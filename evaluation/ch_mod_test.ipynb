```python
def decode(target_text, shift):
    decoded = ""
    base=''
    for ch in target_text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            decoded += chr((ord(ch)-base-shift)%26+base)
        else:
            decoded += ch
    return decoded
```

```python
target_text = "abcdefg"
for i in range(1,10):
    decoded_text = decode(target_text,i)
    print(decoded_text)
```
