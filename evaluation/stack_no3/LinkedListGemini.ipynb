```python
class Node:
    """연결 리스트의 노드 구현"""
    def __init__(self, data):
        self.data = data
        self.next = None
```

2. LinkedList 클래스

```python
class LinkedList:
    """단일 연결 리스트 구현"""
    def __init__(self):
        # 1.1.1. self.head 초기화
        self.head : Node | None = None

    def __len__(self) -> int:
        """1.1.5. 노드의 총 개수를 반환"""
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def insert(self, index, value):
        # 1. index 범위 체크
        if index < 0 or index > len(self):
            raise IndexError("Index out of bounds for insertion")
        
        new_node = Node(value)
        
        # 2. index가 0인 경우: 리스트가 비어 있든 아니든 새로운 head로 지정
        if index == 0:
            new_node.next = self.head  # 기존 head를 next로 연결 (None일 수도 있음)
            self.head = new_node       # 새로운 head로 지정
            return
        
        # 3. index > 0 인 경우: 순회를 통해 이전 노드를 찾아서 연결 변경
        current = self.head
        for _ in range(index - 1):
            current = current.next
        
        new_node.next = current.next
        current.next = new_node

    def delete(self, index):
        """1.1.3. 지정된 index의 노드를 삭제하고 해당 데이터를 반환"""
        if index < 0 or index >= len(self):
            # index 범위 벗어나는 경우 IndexError 처리
            raise IndexError("Index out of bounds for deletion")
            
        if index == 0:
            # head 삭제
            deleted_data = self.head.data
            self.head = self.head.next
            return deleted_data

        # index > 0 인 경우: 삭제할 노드의 이전 노드를 찾기 위해 index - 1 만큼 이동
        current = self.head
        for _ in range(index - 1):
            current = current.next
            
        # current는 삭제할 노드의 이전 노드
        deleted_data = current.next.data
        current.next = current.next.next # 삭제할 노드를 건너뛰고 연결
        return deleted_data

    def to_list(self) -> list:
        """1.1.4. 모든 데이터를 리스트로 묶어서 반환"""
        result = []
        current = self.head
        while current:
            result.append(current.data)
            current = current.next
        return result

    def display(self):
        """1.1.6. (디버깅용) 리스트의 모든 요소를 출력"""
        print("LinkedList:", self.to_list())
```

3. CircularList 클래스

```python
class CircularList:
    """원형 연결 리스트 구현 (last 포인터 사용)"""
    def __init__(self):
        # 1.2.1. self.last 초기화 (cursor 역할)
        self.last = None

    def is_empty(self):
        """리스트가 비어 있는지 확인"""
        return self.last is None
    
    def __len__(self):
        """노드의 총 개수를 반환"""
        if self.is_empty():
            return 0
            
        count = 1
        current = self.last.next # 첫 번째 노드부터 시작
        while current != self.last:
            count += 1
            current = current.next
        return count

    def insert(self, value):
        """1.2.2. 리스트의 끝(last 다음)에 value를 삽입"""
        new_node = Node(value)
        
        if self.is_empty():
            # 리스트가 비어있는 경우
            self.last = new_node
            new_node.next = new_node # 자기 자신을 가리키게 함 (원형)
        else:
            # 리스트가 비어있지 않은 경우: last 노드 다음에 삽입
            new_node.next = self.last.next # 새 노드가 첫 번째 노드를 가리키게
            self.last.next = new_node      # last 노드가 새 노드를 가리키게
            self.last = new_node           # 새 노드가 새로운 last 노드가 됨

    def delete(self, value):
        """1.2.3. 특정 value를 가진 노드를 찾아 삭제하고 해당 데이터를 반환"""
        if self.is_empty():
            raise ValueError("List is empty")
            
        current = self.last.next # 시작 노드 (head)
        previous = self.last     # current의 이전 노드
        
        # 리스트를 순회하며 value를 찾습니다.
        # current가 last.next로 돌아올 때까지 (리스트를 한 바퀴 돌 때까지) 반복
        while True:
            if current.data == value:
                break
            previous = current
            current = current.next
            if current == self.last.next: # 한 바퀴 돌았는데 못 찾은 경우
                raise ValueError(f"Value '{value}' not found in the list")

        # 이제 current는 삭제할 노드, previous는 그 이전 노드입니다.

        if current == self.last and previous == self.last:
            # 리스트에 노드가 하나만 있는 경우
            self.last = None
            return value
        
        # 일반적인 삭제: previous가 current를 건너뛰고 current.next를 가리키게 함
        previous.next = current.next
        
        if current == self.last:
            # 삭제할 노드가 last 노드였던 경우, previous가 새로운 last 노드가 됨
            self.last = previous
        
        return value

    def get_next(self):
        """1.2.4. last를 last.next로 변경하고, 새로운 last의 데이터 (이전 head의 데이터)를 반환"""
        if self.is_empty():
            raise ValueError("List is empty")
            
        # self.last를 다음 노드로 이동 (회전)
        self.last = self.last.next
        
        # 새로운 last 노드의 데이터(이전 head의 데이터) 반환
        return self.last.data

    def search(self, value):
        """1.2.5. value를 가진 노드를 찾아 해당 데이터를 반환"""
        if self.is_empty():
            return None

        current = self.last.next # 시작 노드 (head)
        
        while True:
            if current.data == value:
                return current.data
            current = current.next
            if current == self.last.next: # 한 바퀴 돌았는데 못 찾은 경우
                return None # 못 찾았으면 None 반환

    def display(self):
        """1.2.6. (디버깅용) 리스트의 모든 요소를 출력"""
        if self.is_empty():
            print("CircularList: []")
            return
            
        result = []
        current = self.last.next # head부터 시작
        
        while True:
            result.append(current.data)
            current = current.next
            if current == self.last.next: # head로 다시 돌아오면 종료
                break
                
        print(f"CircularList (Last: {self.last.data}): {result}")
```

4. 테스트: LinkedList

```python
print("## 🧪 LinkedList 테스트 시작 ##")
ll = LinkedList()

# 1. 삽입 테스트 (insert)
ll.insert(0, 10)  # [10]
ll.insert(1, 30)  # [10, 30]
ll.insert(1, 20)  # [10, 20, 30]
ll.insert(3, 40)  # [10, 20, 30, 40]
ll.display()
print(f"길이: {len(ll)}") # 출력: 4

# 2. IndexError 테스트 (삽입)
try:
    ll.insert(5, 50)
except IndexError as e:
    print(f"IndexError (Insert): {e}")

# 3. 삭제 테스트 (delete)
deleted = ll.delete(0) # 10 삭제
print(f"삭제된 데이터 (Index 0): {deleted}") # 출력: 10
ll.display() # [20, 30, 40]

deleted = ll.delete(2) # 40 삭제
print(f"삭제된 데이터 (Index 2): {deleted}") # 출력: 40
ll.display() # [20, 30]

# 4. to_list 테스트
print(f"리스트 변환: {ll.to_list()}") # 출력: [20, 30]

# 5. IndexError 테스트 (삭제)
try:
    ll.delete(2)
except IndexError as e:
    print(f"IndexError (Delete): {e}")
    
print("## 🧪 LinkedList 테스트 종료 ##")
```

5. 테스트: CircularList

```python
print("\n## ⭕ CircularList 테스트 시작 ##")
cl = CircularList()

# 1. 삽입 테스트 (insert)
cl.insert(10) # [10] (last=10)
cl.insert(20) # [10, 20] (last=20)
cl.insert(30) # [10, 20, 30] (last=30)
cl.display()
print(f"길이: {len(cl)}") # 출력: 3

# 2. search 테스트
print(f"Search (30): {cl.search(30)}") # 출력: 30
print(f"Search (99): {cl.search(99)}") # 출력: None

# 3. get_next 테스트 (last 회전 및 데이터 반환)
next_data = cl.get_next() # last는 10이 되고, 10의 데이터 반환
print(f"Next Data: {next_data}") # 출력: 10
cl.display() # [20, 30, 10] (last=10)

# 4. 삭제 테스트 (delete)
deleted = cl.delete(30) # 중간 노드 삭제
print(f"삭제된 데이터 (30): {deleted}") # 출력: 30
cl.display() # [20, 10] (last=10)

deleted = cl.delete(10) # last 노드 삭제
print(f"삭제된 데이터 (10): {deleted}") # 출력: 10
cl.display() # [20] (last=20)

deleted = cl.delete(20) # 마지막 노드 삭제 (리스트 비어짐)
print(f"삭제된 데이터 (20): {deleted}") # 출력: 20
cl.display() # []

# 5. 비어있는 리스트에서의 삭제 테스트
try:
    cl.delete(100)
except ValueError as e:
    print(f"ValueError (Delete on empty list): {e}")
    
# 6. 노드가 하나인 리스트 테스트 후 삭제
cl.insert(5) # [5] (last=5)
cl.display()
cl.delete(5)
cl.display()
    
print("## ⭕ CircularList 테스트 종료 ##")
```
