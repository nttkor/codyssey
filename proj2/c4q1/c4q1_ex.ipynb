3. Transported와 가장 관련성 높은 항목 찾기
Transported 항목은 부울(boolean) 값이므로, 상관관계를 계산하기 전에 숫자로 변환해야 합니다. 그 후 corr() 함수를 사용해 다른 열과의 상관관계를 분석합니다. 

````python

# 전체 데이터 수량 확인
total_rows, total_cols = combined_df.shape
print(f'병합된 전체 데이터의 수량 (행, 열): {combined_df.shape}')

# 1. Transported와 가장 관련성 높은 항목 찾기 (CryoSleep 포함)
# 'Transported' 항목을 숫자로 변환 (True=1, False=0)
combined_df['Transported_numeric'] = combined_df['Transported'].astype(float)

# 'CryoSleep' 항목을 불리언으로 명시적으로 변환 후 NaN을 False로 채우고 정수로 변환하여 경고 해결
combined_df['CryoSleep_numeric'] = combined_df['CryoSleep'].astype('boolean').fillna(False).astype(int)

# 상관관계 계산 (필요한 숫자형 열만 선택)
correlation = combined_df[['Transported_numeric', 'CryoSleep_numeric', 'Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']].corr()['Transported_numeric'].sort_values(ascending=False)
print("Transported 항목과 가장 관련성 높은 항목 (CryoSleep 포함):")
print(correlation)
````

## 2x2 분할표와 결측치 처리
용도: 범주형 변수들 간의 **빈도(Frequency)**를 집계하여 표로 나타냅니다.

````python
# train_df['CryoSleep']의 NaN 값을 False로 채워서 crosstab을 사용했습니다.
train_df['CryoSleep'] = train_df['CryoSleep'].fillna(False)
contingency_table = pd.crosstab(train_df['CryoSleep'], train_df['Transported'])
contingency_table
````

# 4. 나이대별 Transported 여부 시각화

````python
# 4. 나이대별 Transported 여부 시각화
bins = [0, 10, 20, 30, 40, 50, 60, 70, combined_df['Age'].max()]
labels = ['0-9', '10s', '20s', '30s', '40s', '50s', '60s', '70+']
combined_df['AgeGroup'] = pd.cut(combined_df['Age'], bins=bins, labels=labels, right=False)
age_transported = combined_df.groupby(['AgeGroup', 'Transported'])['PassengerId'].count().unstack()
# .unstack()
# unstack()은 MultiIndex의 가장 안쪽 레벨을 행(index)에서 열(column)로 재구성합니다.
# 이 경우, 'Transported'가 가장 안쪽 레벨이므로, Transported의 값들(False, True)이 새로운 열 이름이 됩니다.
# 결과적으로, 'AgeGroup'이 인덱스(행)가 되고, 'Transported'의 값들이 열(column)이 되는 DataFrame이 생성됩니다. 
age_transported.plot(kind='bar', figsize=(12, 7))
plt.title('나이대별 Transported 여부')
plt.xlabel('나이대')
plt.ylabel('승객 수')
plt.xticks(rotation=45)
plt.legend(title='Transported 여부')
plt.show()
````

````python
# 5. Destination별 승객 연령대 분포 시각화
destination_age_distribution = combined_df.groupby(['Destination', 'AgeGroup'])['PassengerId'].count().unstack().fillna(0)
fig, axes = plt.subplots(1, 3, figsize=(20, 6), sharey=True)
destinations = destination_age_distribution.index
for i, destination in enumerate(destinations):
    destination_age_distribution.loc[destination].plot(kind='bar', ax=axes[i], title=f'Destination: {destination}')
    axes[i].set_xlabel('나이대')
    axes[i].set_ylabel('승객 수')
    axes[i].tick_params(axis='x', rotation=45)
plt.suptitle('Destination별 승객 연령대 분포', y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
````

# Transported를 포함해서 수치형과 범주형(라벨 인코딩 후) 컬럼 모두 간의 상관관계를 계산하고 히트맵으로 시각화할 수 있어요.

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1️⃣ 범주형 컬럼 라벨 인코딩
cat_cols = ['HomePlanet', 'CryoSleep', 'Cabin', 'Destination', 'VIP', 'Transported']
df_encoded = df.copy()

for col in cat_cols:
    df_encoded[col] = df_encoded[col].astype(str)  # NaN 처리
    df_encoded[col] = pd.factorize(df_encoded[col])[0]  # 0,1,2,... 숫자로 변환

# 2️⃣ 수치형 컬럼 포함
num_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
cols_for_corr = num_cols + cat_cols

# 3️⃣ 상관계수 계산
corr_matrix = df_encoded[cols_for_corr].corr()

# 4️⃣ 히트맵 시각화
plt.figure(figsize=(12,10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("모든 컬럼 간 상관관계 히트맵")
plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
cat_cols = ['HomePlanet', 'CryoSleep', 'Cabin', 'Destination', 'VIP']

# 수치형 컬럼 구간화
bin_dict = {}
for col in num_cols:
    df[col+'_bin'] = pd.qcut(df[col].fillna(0), q=5, duplicates='drop')
    bin_dict[col] = col+'_bin'

# 시각화 대상 컬럼
plot_cols = cat_cols + list(bin_dict.values())

# Transported True/False 비율 계산
plot_data = []
for col in plot_cols:
    grouped = df.groupby([col, 'Transported'], observed=False).size().reset_index(name='Count')
    grouped_total = grouped.groupby(col, observed=False)['Count'].transform('sum') 
    grouped['Rate'] = grouped['Count'] / grouped_total  # 비율 계산
    grouped['Column'] = col
    plot_data.append(grouped)

plot_df = pd.concat(plot_data, ignore_index=True)

# 시각화
plt.figure(figsize=(15,8))
sns.barplot(
    x='Column',
    y='Rate',
    hue='Transported',
    data=plot_df,
    dodge=True,
    palette={True:'tab:blue', False:'tab:orange'}
)
plt.xticks(rotation=45)
plt.ylabel('Transported 비율')
plt.title('각 컬럼 범주별 Transported True/False 비율')
plt.legend(title='Transported')
plt.show()

````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1️⃣ 데이터 불러오기
# df = pd.read_csv("train.csv")
df = train_df.copy()  # 이미 train_df가 있다고 가정

# 2️⃣ Transported 결측 제거 및 숫자형 변환
df = df.dropna(subset=['Transported'])
df['Transported'] = df['Transported'].astype(int)

# 3️⃣ Age 결측 제거
df = df.dropna(subset=['Age'])

# 4️⃣ AgeGroup 파생 컬럼 생성 (범주형)
bins = [0,10,20,30,40,50,60,70,100]
labels = ['10대 미만','10대','20대','30대','40대','50대','60대','70대 이상']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)

# 5️⃣ 분석용 범주형 컬럼 선택
cat_cols = ['CryoSleep', 'VIP', 'HomePlanet', 'Destination', 'AgeGroup', 'Transported']
df_cat = df[cat_cols].copy()

# 6️⃣ 범주형 → 숫자형 변환 (pandas factorize 사용)
for col in cat_cols:
    df_cat[col] = pd.factorize(df_cat[col])[0]  # 0,1,2,...

# 7️⃣ 상관계수 계산 (절댓값)
corr_cat = df_cat.corr(numeric_only=True).abs()

# 8️⃣ 히트맵 시각화
plt.figure(figsize=(10,8))
sns.heatmap(corr_cat, annot=True, cmap='Reds', fmt=".2f", linewidths=0.5)
plt.title("범주형 컬럼끼리 Transported 상관계수 히트맵", fontsize=15)
plt.show()


````

## 1️⃣ 사용 목적

`sklearn`에서 쓴 이유는 **범주형 데이터를 숫자로 바꾸기 위해서**입니다.

* **범주형 데이터(Categorical)** 예시: `HomePlanet`, `Destination`, `VIP`, `CryoSleep`
* 상관계수(`df.corr()`)는 **숫자형 컬럼**끼리만 계산 가능

따라서, 범주형 컬럼을 **0,1,2,... 같은 숫자형**으로 바꿔주는 과정이 필요합니다.
여기서 사용된 것이:

```python
from sklearn.preprocessing import LabelEncoder
```

* LabelEncoder는 각 범주값을 정수로 변환

  * 예: `['Earth', 'Mars', 'Europa'] → [0,1,2]`

---

## 2️⃣ 요약

| 목적           | 도구                                   |
| ------------ | ------------------------------------ |
| 범주형 → 숫자형 변환 | `sklearn.preprocessing.LabelEncoder` |
| 숫자형 상관계수 계산  | `pandas.DataFrame.corr()`            |
| 머신러닝 모델 학습   | `sklearn` 전반 사용 가능                   |

즉,

> 상관분석 자체에는 `sklearn`이 필요 없지만,
> **범주형 컬럼을 숫자로 바꾸는 전처리** 때문에 사용했던 겁니다.

   * Age, RoomService, FoodCourt, ShoppingMall, Spa, VRDeck

2. **결측치 제거**

   * 수치형 변수에서 NaN이 있으면 상관계수 계산에 방해되므로 제거

3. **절댓값 기준 상관계수 계산**

   * 음수/양수 관계 없이 강도를 시각적으로 확인

4. **Blues 팔레트 히트맵**

   * 값이 클수록 진한 파란색 → Transported와 강한 상관 강조

---

이 히트맵을 보면, **Transported와 가장 관련 있는 수치형 컬럼**이 한눈에 보입니다.

* RoomService, Spa, VRDeck이 음의 상관이 높고, Age는 약간 양의 상관을 보이는 구조를 바로 확인 가능

원하시면 제가 **범주형+수치형을 나눠서 한눈에 비교하는 통합 시각화**도 만들어서 보여드릴 수 있습니다.

그렇게 만들어드릴까요?


````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1️⃣ 데이터 불러오기
# df = pd.read_csv("train.csv")
df = train_df.copy()  # 이미 train_df가 있다고 가정

# 2️⃣ Transported 결측 제거 및 숫자형 변환
df = df.dropna(subset=['Transported'])
df['Transported'] = df['Transported'].astype(int)

# 3️⃣ Age 및 수치형 컬럼 결측 제거 (필요 시)
num_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
df = df.dropna(subset=num_cols)

# 4️⃣ 분석용 수치형 컬럼 선택
selected_cols = ['Transported'] + num_cols
df_num = df[selected_cols].copy()

# 5️⃣ 상관계수 계산 (절댓값 기준)
corr_num = df_num.corr(numeric_only=True).abs()

# 6️⃣ 히트맵 시각화
plt.figure(figsize=(8,6))
sns.heatmap(corr_num, annot=True, cmap='Blues', fmt=".2f", linewidths=0.5)
plt.title("Transported 관련 수치형 변수 상관계수 히트맵", fontsize=15)
plt.show()
````

## 4️⃣ 시각화로 확인 (범주형 변수 포함)

범주형 변수는 단순 상관계수로는 불충분하므로 **막대그래프나 교차표 시각화**로 확인합니다.

````python
# =====================================
# 🚀 나이대별 Transported 비율 시각화
# =====================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from matplotlib import rc

# 1️⃣ 시스템에 설치된 한글 폰트 확인
# 아래 코드를 실행하면 한글 지원 폰트 목록이 나옵니다.
# [참고] Ubuntu/Linus는 'NanumGothic'이 가장 무난합니다.
# fm.findSystemFonts(fontpaths=None, fontext='ttf')

# 2️⃣ 폰트 설정 (NanumGothic 예시)
plt.rcParams['font.family'] = 'NanumGothic'

# 3️⃣ 마이너스 부호 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False


# 4️⃣ Age 결측 제거
df = df.dropna(subset=['Age'])

# 5️⃣ 연령대 구간 만들기 (10살 단위)
bins = [0, 10, 20, 30, 40, 50, 60, 70, 100]
labels = ['10세 미만', '10대', '20대', '30대', '40대', '50대', '60대', '70세 이상']
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)

# 6️⃣ 연령대별 Transported 비율 계산
age_group_mean = df.groupby('AgeGroup', observed=False)['Transported'].mean().reset_index()

# 7️⃣ 시각화
plt.figure(figsize=(10,6))
sns.barplot(
    x='AgeGroup', 
    y='Transported', 
    data=age_group_mean,
    dodge=False,  # hue가 없으므로 도지 불필요
    palette='coolwarm'
)

plt.title('연령대별 Transported 비율', fontsize=16)
plt.xlabel('연령대', fontsize=12)
plt.ylabel('Transported 비율 (0~1)', fontsize=12)
plt.ylim(0, 1)  # 비율이므로 0~1 범위
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1️⃣ 범주형 컬럼 라벨 인코딩
cat_cols = ['HomePlanet', 'CryoSleep', 'Cabin', 'Destination', 'VIP', 'Transported']
df_encoded = df.copy()

for col in cat_cols:
    df_encoded[col] = df_encoded[col].astype(str)  # NaN 처리
    df_encoded[col] = pd.factorize(df_encoded[col])[0]  # 0,1,2,... 숫자로 변환

# 2️⃣ 수치형 컬럼 포함
num_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
cols_for_corr = num_cols + cat_cols

# 3️⃣ 상관계수 계산
corr_matrix = df_encoded[cols_for_corr].corr()

# 4️⃣ 히트맵 시각화
plt.figure(figsize=(12,10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("모든 컬럼 간 상관관계 히트맵")
plt.show()
````
