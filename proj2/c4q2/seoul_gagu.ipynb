정규식 설명
정규식 r'^\d{4}(\.\d+)?$'는 문자열 전체가 네 자리 숫자(예: 2015)이거나 네 자리 숫자 뒤에 점과 하나 이상의 숫자(예: 2015.1, 2024.19)가 선택적으로 올 수 있음을 의미한다.

^: 문자열 시작을 뜻한다.

\d{4}: 숫자(digit) 4개가 정확히 연속해서 나옴을 뜻한다.

(\.\d+)?: 소수점 문자('.')와 하나 이상의 숫자(\d+)로 구성된 그룹이 0번 또는 1번 나타날 수 있음을 뜻한다(옵션).

$: 문자열 끝을 뜻한다.

예시 매칭: "2015" → 매칭, "2015.2" → 매칭, "15.1" → 불일치, "2015." → 불일치, "2015.2a" → 불일치

```python
# df: '구' 컬럼은 이미 drop 됨
# 예: df.columns => ['동별(1)', '동별(2)', '2015', '2015.1', '2015.2', ..., '2024.19']
import re
# 1) 연도 패턴 매칭: 'YYYY' 또는 'YYYY.x' 형태 컬럼 선택
year_cols = [c for c in df.columns if re.match(r'^\d{4}(\.\d+)?$', c)]
```

```python
", ".join(word for word in year_cols[:10])
```

```python
# 2) 연도 키 추출 (ex: '2015.2' -> '2015')
col_year_map = {c: re.match(r'^(\d{4})', c).group(1) for c in year_cols}
", ".join(f'{key}:{items}' for key,items in list(col_year_map.items())[:10])
```

```python
# 3) 같은 연도끼리 합산: 먼저 연도별로 빈 프레임 만들기
years = sorted({v for v in col_year_map.values()})  # 중복을 제거한 년도 만들기
year_df = pd.DataFrame(index=df.index) #먼저 연도별로 빈 프레임 만들기
```

```python
for y in years:
    cols_for_y = [col for col, yy in col_year_map.items() if yy == y]
    # NaN 안전하게 합: sum(axis=1, skipna=True)
    year_df[y] = df[cols_for_y].sum(axis=1)
```

```python
# 4) 기존의 연도 관련 상세 컬럼 제거하고 연도 합계 붙이기
#    (원하면 동별 식별자 컬럼은 유지)
non_year_cols = [c for c in df.columns if c not in year_cols]
result = pd.concat([df[non_year_cols].reset_index(drop=True), year_df.reset_index(drop=True)], axis=1)
```

```python
# 5) 연도 컬럼을 정수형으로 정리 (컬럼명으로 사용)
#    현재 year_df의 컬럼은 문자열 '2015' 등임. 필요시 int로 변환 가능:
result.columns = [int(c) if re.match(r'^\d{4}$', str(c)) else c for c in result.columns]

# 결과 확인
print(result.head())
```

```python
# 2) 연도 관련 컬럼 선택(예: '2015', '2015.1', ..., '2024.19' 등 숫자로 시작하는 컬럼)
year_cols = [c for c in df.columns if re.match(r'^\d{4}(\.\d+)?$', c)]
```

```python
# 3) melt로 와이드 -> 롱 변환
#    variable_name은 원래 컬럼 이름(예: '2015.1')을 담고, value_name은 해당 값
long = df.melt(id_vars=[c for c in df.columns if c not in year_cols],
               value_vars=year_cols,
               var_name='year_col',
               value_name='value')
```

```python
# 4) 필요하면 'year_col'에서 실제 연도만 추출 (예: '2015.1' -> '2015' 또는 '2015.1'을 유지)
#    (아래는 소수점 앞 숫자만 연도로 추출하는 예)
long['year'] = long['year_col'].str.extract(r'^(\d{4})').astype(int)
```

```python
# 5) 불필요한 컬럼 제거 및 정렬
long = long.drop(columns=['year_col'])
long = long.reset_index(drop=True)

# 결과 확인
print(long.head())
```

groupby로 간단히 합치기

```python
import pandas as pd 
import re
df = pd.read_csv('seoul_gagu.csv')
df.size

# 가정: df는 원본 DataFrame (동/구 컬럼은 이미 제거하거나 유지할 수 있음)
df = df.drop(columns=['동별(1)', '동별(2)'], errors='ignore')
# 1) 컬럼명에서 연도(앞 4자리) 추출 (매칭되지 않으면 NaN)
col_year = df.columns.to_series().str.extract(r'^(\d{4})')[0]

# 2) 연도 형태로 매칭된 컬럼들만 같은 연도끼리 합산
df_year = df.loc[:, col_year.notna()].groupby(col_year[col_year.notna()], axis=1).sum()

# 3) 연도 외(식별자 등) 컬럼이 있다면 합쳐서 최종 결과 생성
non_year = df.loc[:, col_year.isna()]
result = pd.concat([non_year.reset_index(drop=True), df_year.reset_index(drop=True)], axis=1)

# 4) (선택) 연도 컬럼을 정수형으로 바꾸려면
result.columns = [int(c) if isinstance(c, str) and c.isdigit() else c for c in result.columns]


```
